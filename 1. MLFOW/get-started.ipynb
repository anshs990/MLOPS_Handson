## MLFLow Tracking Server

```python
import mlflow
```

```python
mlflow.set_tracking_uri("http://localhost:5000")
```

```python
mlflow.set_experiment("Check localhost connection")

with mlflow.start_run():
    mlflow.log_param("param1", 5)
    mlflow.log_metric("metric1", 0.89)
    mlflow.log_param("param2", "test")
    mlflow.log_metric("metric2", 0.95)
```

```python
with mlflow.start_run():
    mlflow.log_metric("test1",1)
    mlflow.log_metric("Krish1",2)
```

```python
with mlflow.start_run():
    mlflow.log_metric("test2",1)
    mlflow.log_metric("Krish2",2)
```

```python
# Same experiment, multiple runs
mlflow.set_experiment("My Experiment")

# Run 1
with mlflow.start_run(run_name="Run 1"):
    mlflow.log_param("param1", 5)
    mlflow.log_metric("metric1", 0.89)

# Run 2  
with mlflow.start_run(run_name="Run 2"):
    mlflow.log_param("param2", "test")
    mlflow.log_metric("metric2", 0.95)

# Run 3
with mlflow.start_run(run_name="Run 3"):
    mlflow.log_metric("test1", 1)

# Run 4
with mlflow.start_run(run_name="Run 4"):
    mlflow.log_metric("Krish1", 2)
```
